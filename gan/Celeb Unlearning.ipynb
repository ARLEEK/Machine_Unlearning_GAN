```python
import re

pattern = re.compile(
    r"Epoch (\d+): Generated (\d+) images\. With Glasses: (\d+) \(([\d.]+)%\)\. Kept \(Clean\): (\d+).*?"
    r"D Loss: ([\d\-.]+) G Loss: ([\d\-.]+) \(Anchor: ([\d\-.]+)\)",
    re.DOTALL
)

results = []

for match in pattern.finditer(output_string):
    epoch, total, glasses, pct, kept, dloss, gloss, anchor = match.groups()
    results.append({
        "epoch": int(epoch),
        "generated": int(total),
        "glasses_count": int(glasses),
        "glasses_pct": float(pct),
        "kept": int(kept),
        "d_loss": float(dloss),
        "g_loss": float(gloss),
        "anchor_loss": float(anchor),
    })

results[:5]  # first 5 epochs preview
```

```python
import matplotlib.pyplot as plt

epochs = [r["epoch"] for r in results]
glass_prob = [r["glasses_pct"] for r in results]

plt.figure(figsize=(8,5))
plt.plot(epochs, glass_prob)
plt.xlabel("Epoch")
plt.ylabel("Probability of Glasses (%)")
plt.title("Probability of Seeing Glasses Over Epochs")
plt.grid(True)
plt.show()
```
